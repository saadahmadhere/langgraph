### Pydantic Data Validation

```python
from langgraph.graph import StateGraph,START,END
from pydantic import BaseModel
```

```python
class State(BaseModel):
    name:str
    
```

```python
## node function
def example_node(state:State):
    return {"name":"Hello"}
```

```python
## stateGraph
builder=StateGraph(State)
builder.add_node("example node",example_node)

builder.add_edge(START,"example node")
builder.add_edge("example node",END)

graph=builder.compile()
```

```python
graph.invoke({"name":"saad"})
```
